```python
# Inspect CSV Columns Efficiently

# This notebook reads only the header row of a very large CSV file
# so it can list the column names without loading the full dataset.

import pandas as pd
from pathlib import Path

# Path to your CSV file
csv_path = Path("csv_output/gaming_r0_20251225.csv")

if not csv_path.exists():
    raise FileNotFoundError(f"File not found: {csv_path}")

print(f"Inspecting: {csv_path}")
print(f"File size: {csv_path.stat().st_size / (1024**3):.2f} GB")

# Read only the header row
try:
    df_header = pd.read_csv(csv_path, nrows=0)
    columns = list(df_header.columns)
except Exception as e:
    print(f"Pandas could not read the header directly: {e}")
    raise

print(f"Number of columns: {len(columns)}")
print("Columns in the dataset:")
for i, col in enumerate(columns, start=1):
    print(f"{i:>3}. {col}")

# Read first 5 rows (small sample, safe even for large file)
print("First 5 rows of the dataset:")
try:
    df_sample = pd.read_csv(csv_path, nrows=5)
    display(df_sample)
except Exception as e:
    print(f"Error reading first rows: {e}")
```
